#### Best practices for handling geospatial data

1. **Always verify the CRS** before any operation — mixing EPSG:4326 and EPSG:27700 silently gives wrong results
2. **Keep raw data immutable** — never overwrite source files; write outputs to a separate `output/` folder
3. **Validate coordinates** before loading — lat ∈ [−90, 90], lon ∈ [−180, 180]
4. **Use UTF-8 encoding** when reading text-based geodata (GeoJSON, CSV with city names)
5. **Log data quality issues** — missing values, out-of-range coords, duplicate IDs
6. **Document provenance** — record where the data came from, when it was downloaded, and what version it is

## Section 2 — Python Syntax and Indentation

Python is designed for human readability. Its most important design choice:
**indentation defines code blocks**, not curly braces.

```
JavaScript (uses braces)      Python (uses indentation)
──────────────────────────    ─────────────────────────
if (valid) {                  if valid:
    process();                    process()
}                             
```

### Three golden rules

| Rule | Good ✅ | Bad ❌ |
|------|---------|--------|
| 4-space indentation | `    return x` | `  return x` or `	return x` |
| One statement per line | `x = 1` | `x = 1; y = 2` |
| Comments explain *why* | `# degrees → radians for trig` | `# multiply by pi/180` |

> 💡 **Never mix tabs and spaces.** Configure your editor to use 4 spaces per indent level.
> In VS Code: set `"editor.tabSize": 4` and `"editor.insertSpaces": true`.

````python
# 💻 2.1  Your first Python cells — syntax and indentation

# Single-line comment: Python ignores everything after #
print("Hello, Geospatial World!")   # inline comment

# ── Indentation example ──────────────────────────────────────
city       = "Amsterdam"
population = 921_402     # underscores make large numbers readable

if population > 1_000_000:            # colon starts a new block
    print(f"{city} is a large city")  # ← 4-space indent
    print("Population: over 1 million")
elif population > 500_000:
    print(f"{city} is a medium city") # ← this block executes
    print(f"Population: {population:,}")
else:
    print(f"{city} is a small city")

print("City classification complete.")  # ← back at top level (no indent)
````

## Section 3 — Python Data Types

```
Python built-in types used in geospatial work
│
├── int      →  42, -86, 4326           (IDs, EPSG codes, floor counts)
├── float    →  51.5074, -0.1278        (coordinates, distances, areas)
├── str      →  "London", "EPSG:4326"  (names, WKT, file paths)
├── bool     →  True / False            (validity flags, filter masks)
└── None     →  None                   (missing attributes, optional values)
```

> 💡 In geospatial work you will encounter all five types within a single CSV row:
> `id` (int), `lat/lon` (float), `name` (str), `is_heritage` (bool), `elevation` (None if not recorded).

````python
# 💻 3.1  int — whole numbers

feature_id   = 1042          # unique identifier
epsg_code    = 4326          # coordinate reference system code
floor_count  = 24            # floors in a building
population   = 9_083_522     # city population (underscore = thousands separator)
elevation_m  = -86           # elevation (negative = below sea level)

print("=== Integer examples ===")
print(type(feature_id))            # <class 'int'>
print(f"Feature ID   : {feature_id}")
print(f"EPSG code    : {epsg_code}")
print(f"Population   : {population:,}")   # comma-formatted
print(f"Elevation    : {elevation_m} m")

# Integers support all arithmetic operators
print(f"
5 ** 2 = {5 ** 2}")      # exponentiation
print(f"17 // 5 = {17 // 5}")     # integer (floor) division → 3
print(f"17  % 5 = {17  % 5}")     # modulo (remainder) → 2
````

````python
# 💻 3.5  Type checking and safe conversion
#
# In real-world geospatial ETL pipelines, EVERY value arrives as a string
# from a CSV, JSON, or database. You must convert and validate before use.

raw_lat       = "51.5074"    # came in as text from a CSV
raw_pop       = "9083522"
raw_flag      = "1"
raw_elevation = "N/A"        # common sentinel for missing numeric data

print("=== type() checks ===")
print(f"raw_lat  → {type(raw_lat).__name__}")
print(f"raw_pop  → {type(raw_pop).__name__}")

print("
=== Safe conversion ===")
lat_float  = float(raw_lat)              # "51.5074" → 51.5074
pop_int    = int(raw_pop)                # "9083522" → 9083522
flag_bool  = raw_flag == "1"             # string comparison → bool
print(f"lat_float  = {lat_float}   ({type(lat_float).__name__})")
print(f"pop_int    = {pop_int:,}  ({type(pop_int).__name__})")
print(f"flag_bool  = {flag_bool}    ({type(flag_bool).__name__})")

# Graceful handling of unparseable values
try:
    elevation = float(raw_elevation)     # will raise ValueError
except ValueError:
    elevation = None
    print(f"
Could not convert '{raw_elevation}' → stored as None")

print(f"elevation  = {elevation}")
````

## Section 4 — Variables and Naming Conventions

### Hard rules (break these → SyntaxError)
- Must start with a letter or `_`: `lat` ✅, `_raw` ✅, `1lat` ❌
- Letters, digits, underscores only: `lat_deg_2` ✅, `lat-deg` ❌
- Case-sensitive: `Lat` ≠ `lat` ≠ `LAT`
- Cannot shadow reserved words: `if`, `for`, `class`, `return`, `None`, `True`, `False` …

### PEP 8 style conventions (break these → ugly code)

| Item | Convention | Example |
|------|-----------|---------|
| Variables & functions | `snake_case` | `city_name`, `parse_coordinate()` |
| Constants | `UPPER_SNAKE_CASE` | `EARTH_RADIUS_KM = 6371` |
| Classes | `PascalCase` | `GeoFeature`, `SpatialIndex` |
| "Private" (internal) | `_leading_underscore` | `_raw_cache` |
| Module-level singletons | `UPPER_SNAKE_CASE` | `DEFAULT_CRS = "EPSG:4326"` |

````python
# 💻 4.1  Variable assignment patterns

# ── Constants (never change after definition) ────────────────
EARTH_RADIUS_KM  = 6_371.0
WGS84_EPSG       = 4326
DEFAULT_CRS      = "EPSG:4326"
MAX_LAT, MIN_LAT = 90.0, -90.0
MAX_LON, MIN_LON = 180.0, -180.0

# ── Multiple assignment ──────────────────────────────────────
x, y, z = 10.5, 20.3, 0.0          # unpack tuple → three variables

# ── Variable swap (no temp variable needed) ──────────────────
lat, lon = 48.8566, 2.3522          # Paris
print(f"Before swap: lat={lat}, lon={lon}")
lat, lon = lon, lat
print(f"After swap : lat={lat}, lon={lon}")

# ── Augmented assignment ─────────────────────────────────────
distance_km = 100.0
distance_km += 50.5                 # distance_km = distance_km + 50.5
distance_km *= 1.1                  # apply 10% correction factor
print(f"
Final distance: {distance_km:.2f} km")

# ── Type hints (Python 3.5+, strongly recommended) ──────────
# Hints don't enforce types at runtime, but document intent
# and enable IDE autocomplete + static analysis (mypy)
latitude:  float = 51.5074
city_name: str   = "London"
feature_count: int = 0
is_loaded: bool  = False
print(f"
Latitude : {latitude}  ({type(latitude).__name__})")
````

## Section 5 — Operators

| Category | Operators | Notes |
|----------|-----------|-------|
| **Arithmetic** | `+  -  *  /  //  %  **` | `//` = floor div, `%` = modulo, `**` = power |
| **Comparison** | `==  !=  <  >  <=  >=` | Returns `bool` |
| **Logical** | `and  or  not` | Short-circuit evaluation |
| **Identity** | `is  is not` | Object identity (use for `None` checks) |
| **Membership** | `in  not in` | Works on strings, lists, dicts, sets |

````python
# 💻 5.1  Arithmetic operators — geospatial calculations
import math

lat1, lon1 =  51.5074, -0.1278   # London
lat2, lon2 =  48.8566,  2.3522   # Paris

delta_lat = lat2 - lat1
delta_lon = lon2 - lon1
print(f"Δ latitude  : {delta_lat:+.4f}°")
print(f"Δ longitude : {delta_lon:+.4f}°")

# DMS conversion with // and %
total_seconds = 185_432            # coordinate in arc-seconds
degrees   = total_seconds // 3600
remainder = total_seconds  % 3600
minutes   = remainder      // 60
seconds   = remainder       % 60
print(f"
{total_seconds} arc-sec = {degrees}° {minutes}' {seconds}"")

# Exponentiation — used in Haversine distance formula
dx_sq = (delta_lon * math.cos(math.radians((lat1 + lat2) / 2))) ** 2
dy_sq = delta_lat ** 2
approx_km = math.sqrt(dx_sq + dy_sq) * 111.32     # 1° ≈ 111.32 km
print(f"
Approx London–Paris (equirectangular): {approx_km:.1f} km")
````

````python
# 💻 7.1  Defining and calling functions

def describe_point(lat: float, lon: float, label: str = "Location") -> str:
    """Format a WGS84 coordinate as a human-readable string.

    Args:
        lat: Latitude in decimal degrees (-90 to 90).
        lon: Longitude in decimal degrees (-180 to 180).
        label: Optional place label. Defaults to "Location".

    Returns:
        Formatted string like "London: 51.5074°N, 0.1278°W"
    """
    lat_dir = "N" if lat >= 0 else "S"
    lon_dir = "E" if lon >= 0 else "W"
    return f"{label}: {abs(lat):.4f}°{lat_dir}, {abs(lon):.4f}°{lon_dir}"

# Positional args
print(describe_point(51.5074, -0.1278))

# Keyword args (order doesn't matter)
print(describe_point(lon=2.3522, lat=48.8566, label="Paris"))

# Default parameter (label omitted)
print(describe_point(-33.8688, 151.2093))
````

````python
# 💻 7.2  The Haversine formula — a real geospatial function
import math

def haversine_km(lat1: float, lon1: float,
                 lat2: float, lon2: float) -> float:
    """Calculate the great-circle distance between two WGS84 points.

    Uses the Haversine formula. Accurate to within ~0.3% for most distances.

    Args:
        lat1, lon1: Start point, decimal degrees.
        lat2, lon2: End point, decimal degrees.

    Returns:
        Distance in kilometres (float).
    """
    R = 6_371.0                              # Earth mean radius in km
    phi1, phi2 = math.radians(lat1), math.radians(lat2)
    dphi       = math.radians(lat2 - lat1)
    dlambda    = math.radians(lon2 - lon1)

    a = (math.sin(dphi / 2) ** 2
         + math.cos(phi1) * math.cos(phi2) * math.sin(dlambda / 2) ** 2)
    return R * 2 * math.atan2(math.sqrt(a), math.sqrt(1 - a))


# Test with known city pairs
pairs = [
    ("London",   51.5074, -0.1278, "Paris",   48.8566,  2.3522),
    ("London",   51.5074, -0.1278, "New York", 40.7128, -74.0060),
    ("Sydney",  -33.8688, 151.2093, "Tokyo",   35.6762,  139.6503),
]
print(f"{'Route':<30} {'Distance':>12}")
print("─" * 44)
for a, la, lo, b, lb, lob in pairs:
    d = haversine_km(la, lo, lb, lob)
    print(f"{a} → {b:<18} {d:>10,.1f} km")
````

````python
# 💻 7.3  Multiple return values, *args, **kwargs

def bounding_box(coords: list) -> tuple:
    """Return the (min_lat, max_lat, min_lon, max_lon) of a coordinate list.

    Returns:
        Tuple (min_lat, max_lat, min_lon, max_lon).

    Raises:
        ValueError: If coords is empty.
    """
    if not coords:
        raise ValueError("Cannot compute bounding box of empty list.")
    lats = [lat for lat, _ in coords]
    lons = [lon for _, lon in coords]
    return min(lats), max(lats), min(lons), max(lons)

# Unpack multiple return values
europe = [(51.5, -0.1), (48.9, 2.4), (52.5, 13.4), (41.9, 12.5), (40.4, -3.7)]
min_lat, max_lat, min_lon, max_lon = bounding_box(europe)
print(f"Bounding box:  lat [{min_lat:.1f}, {max_lat:.1f}]  lon [{min_lon:.1f}, {max_lon:.1f}]")

# ── **kwargs for flexible metadata logging ───────────────────
def log_event(event: str, **metadata) -> None:
    """Print a structured log entry."""
    parts = [event] + [f"{k}={v!r}" for k, v in metadata.items()]
    print(f"[LOG] {' | '.join(parts)}")

log_event("file_loaded", path="cities.geojson", rows=1402)
log_event("transform",   src_crs="EPSG:4326", dst_crs="EPSG:27700", duration_ms=32)
````

````python
# 💻 8.1  String cleaning — the most common ETL task

# Raw values with real-world messiness
raw_city = "  new york city  "
raw_crs  = "  EPSG : 4326  "
raw_wkt  = "POINT( 2.3522  48.8566 )"

# ── Basic cleaning ───────────────────────────────────────────
clean_city = raw_city.strip().title()              # "New York City"
clean_crs  = raw_crs.strip().replace(" ", "").upper()   # "EPSG:4326"
print(f"city : {clean_city!r}")
print(f"crs  : {clean_crs!r}")

# ── Parse WKT coordinates ────────────────────────────────────
# "POINT( 2.3522  48.8566 )" → lon=2.3522, lat=48.8566
inner  = raw_wkt[raw_wkt.find("(") + 1 : raw_wkt.find(")")]
parts  = inner.split()                             # split on whitespace
lon, lat = float(parts[0]), float(parts[1])
print(f"
Parsed WKT: lon={lon}, lat={lat}")

# ── Checking file types ──────────────────────────────────────
filenames = ["cities.geojson", "buildings.gpkg", "dem.tif", "notes.txt"]
geo_exts  = {".geojson", ".gpkg", ".shp", ".tif", ".tiff", ".csv"}
for fn in filenames:
    ext   = fn[fn.rfind("."):]
    label = "🗺️ geo" if ext in geo_exts else "📄 other"
    print(f"  {fn:<22}  ext={ext:<9}  {label}")
````

## Section 10 — Error Handling

Real-world geospatial data is messy. Robust code **anticipates failure** and handles
it gracefully instead of crashing.

### Common exceptions in geospatial work

| Exception | When you see it |
|-----------|----------------|
| `ValueError` | `float("N/A")` — wrong value for the type |
| `TypeError` | `len(42)` — wrong argument type |
| `KeyError` | `d["missing_key"]` — dict key absent |
| `IndexError` | `lst[9999]` — index out of range |
| `FileNotFoundError` | Opening a file that does not exist |
| `AssertionError` | An `assert` condition evaluated to `False` |

### The try / except / else / finally pattern

```python
try:
    result = risky_operation()      # code that might fail
except ValueError as e:
    result = fallback               # handle specific error
except (TypeError, KeyError) as e:
    result = None                   # handle multiple errors
else:
    post_process(result)            # runs ONLY if no exception
finally:
    cleanup()                       # ALWAYS runs (even on exception)
```

````python
# 💻 10.1  try / except — safe coordinate parsing

def parse_float(value, field_name: str = "value") -> float | None:
    """Safely convert a value to float; returns None on failure.

    Args:
        value: The raw value (str, int, float, or None).
        field_name: Used in warning messages.

    Returns:
        float, or None if conversion fails.
    """
    if value is None:
        return None
    try:
        return float(str(value).strip())
    except (ValueError, TypeError) as e:
        print(f"  ⚠️  Cannot parse {field_name}={value!r}: {e}")
        return None


# Test with a variety of inputs
test_cases = [
    ("51.5074", "lat"),
    ("-0.1278", "lon"),
    ("N/A",     "lat"),
    ("",        "lon"),
    (None,      "lat"),
    ("51°30'",  "lat"),     # DMS — would need separate parser
    (51.5,      "lat"),     # already a float — should still work
]

print("parse_float results:")
for val, field in test_cases:
    result = parse_float(val, field)
    print(f"  {str(val):15s} → {result}")
````

````python
# 💻 10.2  assert — document and enforce preconditions

import math

def haversine_km_safe(lat1, lon1, lat2, lon2) -> float:
    """Calculate great-circle distance with precondition checks.

    Raises:
        AssertionError: If any coordinate is outside valid WGS84 range.
    """
    assert -90  <= lat1 <=  90,  f"lat1 out of range: {lat1}"
    assert -90  <= lat2 <=  90,  f"lat2 out of range: {lat2}"
    assert -180 <= lon1 <= 180,  f"lon1 out of range: {lon1}"
    assert -180 <= lon2 <= 180,  f"lon2 out of range: {lon2}"

    R = 6_371.0
    phi1, phi2 = math.radians(lat1), math.radians(lat2)
    dphi       = math.radians(lat2 - lat1)
    dlambda    = math.radians(lon2 - lon1)
    a = math.sin(dphi/2)**2 + math.cos(phi1)*math.cos(phi2)*math.sin(dlambda/2)**2
    return R * 2 * math.atan2(math.sqrt(a), math.sqrt(1-a))

print(f"London → Paris: {haversine_km_safe(51.5074, -0.1278, 48.8566, 2.3522):.1f} km")

# This will raise AssertionError — caught here to demonstrate
try:
    haversine_km_safe(200.0, -0.1278, 48.8566, 2.3522)
except AssertionError as e:
    print(f"
AssertionError: {e}")
````

````python
# 💻 11.1  PEP 8 before / after comparison
import math

# ─── BEFORE (hard to read) ───────────────────────────────────
def calc(a,b,c,d):
    R=6371
    p1=math.radians(a);p2=math.radians(c)
    dp=math.radians(c-a);dl=math.radians(d-b)
    x=math.sin(dp/2)**2+math.cos(p1)*math.cos(p2)*math.sin(dl/2)**2
    return R*2*math.atan2(math.sqrt(x),math.sqrt(1-x))

# ─── AFTER (PEP 8 + docstring + type hints) ──────────────────
def haversine_km(lat1: float, lon1: float,
                 lat2: float, lon2: float) -> float:
    """Calculate great-circle distance between two WGS84 points.

    Args:
        lat1, lon1: Start point, decimal degrees.
        lat2, lon2: End point, decimal degrees.

    Returns:
        Distance in kilometres.

    Example:
        >>> haversine_km(51.5074, -0.1278, 48.8566, 2.3522)
        343.06...
    """
    EARTH_RADIUS_KM = 6_371.0
    phi1 = math.radians(lat1)
    phi2 = math.radians(lat2)
    dphi    = math.radians(lat2 - lat1)
    dlambda = math.radians(lon2 - lon1)
    a = (math.sin(dphi / 2) ** 2
         + math.cos(phi1) * math.cos(phi2) * math.sin(dlambda / 2) ** 2)
    return EARTH_RADIUS_KM * 2 * math.atan2(math.sqrt(a), math.sqrt(1 - a))

# Both return the same result
d_old = calc(51.5074, -0.1278, 48.8566, 2.3522)
d_new = haversine_km(51.5074, -0.1278, 48.8566, 2.3522)
print(f"Old: {d_old:.2f} km")
print(f"New: {d_new:.2f} km")
print("\nThe 'after' version is self-documenting, testable, and maintainable.")
````

````python
# 🔬 Step 2 — Low-level helper functions (one job each)

def clean_string(value, field_name: str = "field") -> str | None:
    """Strip whitespace from a string; return None if empty/None.

    Args:
        value: Raw string value (may be None).
        field_name: Used in debug messages.

    Returns:
        Stripped string, or None if blank or None.
    """
    if value is None:
        return None
    cleaned = str(value).strip()
    return cleaned if cleaned else None


def clean_coordinate(value, field_name: str = "coord") -> float | None:
    """Parse a raw value to float coordinate; return None on failure.

    Args:
        value: Raw coordinate (str, float, int, or None).
        field_name: Used in debug messages.

    Returns:
        float, or None if conversion fails.
    """
    if value is None:
        return None
    stripped = str(value).strip()
    if not stripped:
        return None
    try:
        return float(stripped)
    except ValueError:
        return None


def validate_lat_lon(lat: float | None, lon: float | None) -> bool:
    """Return True only if both lat and lon are non-None and in WGS84 range."""
    if lat is None or lon is None:
        return False
    return -90 <= lat <= 90 and -180 <= lon <= 180


# ── Inline unit tests ─────────────────────────────────────────
assert clean_string("  London  ")    == "London"
assert clean_string(None)            is None
assert clean_string("  ")            is None
assert clean_coordinate("51.5074")   == 51.5074
assert clean_coordinate("  -0.12 ")  == -0.12
assert clean_coordinate("N/A")       is None
assert clean_coordinate(None)        is None
assert clean_coordinate("")          is None
assert validate_lat_lon(51.5, -0.1)  is True
assert validate_lat_lon(200.0, 0.0)  is False
assert validate_lat_lon(None, 0.0)   is False

print("✅ All helper function unit tests passed!")
````

````python
# 🔬 Step 3 — The main record-cleaning function

VALID_STATUSES = {"active", "inactive", "maintenance"}
MIN_NAME_LENGTH = 3           # names shorter than this are rejected

def clean_hotspot_record(raw: tuple) -> dict | None:
    """Parse and validate one raw Wi-Fi hotspot record.

    Args:
        raw: Tuple of (id, name, lat_raw, lon_raw, status, city).

    Returns:
        Cleaned dict, or None if the record fails validation.
    """
    feature_id, raw_name, raw_lat, raw_lon, raw_status, raw_city = raw

    # ── Clean each field ─────────────────────────────────────
    name   = clean_string(raw_name,   "name")
    lat    = clean_coordinate(raw_lat, "lat")
    lon    = clean_coordinate(raw_lon, "lon")
    status = clean_string(raw_status, "status")
    city   = clean_string(raw_city,   "city")

    # ── Normalise values ─────────────────────────────────────
    if city:
        city = city.title()            # "LONDON" → "London"
    if status:
        status = status.lower()        # "Active"  → "active"

    # ── Validate required fields ─────────────────────────────
    if name is None:
        print(f"  SKIP  id={feature_id:>2}: missing name"); return None

    if len(name) < MIN_NAME_LENGTH:
        print(f"  SKIP  id={feature_id:>2}: name '{name}' too short (< {MIN_NAME_LENGTH} chars)"); return None

    if not validate_lat_lon(lat, lon):
        print(f"  SKIP  id={feature_id:>2} '{name}': invalid coords (lat={lat}, lon={lon})"); return None

    if city is None:
        print(f"  SKIP  id={feature_id:>2} '{name}': missing city"); return None

    if status not in VALID_STATUSES:
        print(f"  SKIP  id={feature_id:>2} '{name}': unknown status '{status}'"); return None

    return {"id": feature_id, "name": name, "lat": lat,
            "lon": lon, "status": status, "city": city}


# Quick smoke-test
print("Testing clean_hotspot_record:
")
print("  Record 1 (good) →", clean_hotspot_record(RAW_HOTSPOTS[0]))
print()
print("  Record 6 (bad lat) →", clean_hotspot_record(RAW_HOTSPOTS[5]))
print()
print("  Record 13 (None name) →", clean_hotspot_record(RAW_HOTSPOTS[12]))
````

````python
# 🔬 Step 4 — Deduplication

def deduplicate(records: list[dict]) -> list[dict]:
    """Remove exact duplicate records based on (name, lat, lon).

    Keeps the first occurrence; silently drops subsequent duplicates
    with a printed message.

    Args:
        records: List of cleaned record dicts.

    Returns:
        List with duplicates removed.
    """
    seen    = set()
    unique  = []
    for rec in records:
        # Round coordinates to 6 decimal places to avoid float noise
        key = (rec["name"].lower(),
               round(rec["lat"], 6),
               round(rec["lon"], 6))
        if key in seen:
            print(f"  DEDUP id={rec['id']:>2}: removing duplicate '{rec['name']}'")
            continue
        seen.add(key)
        unique.append(rec)
    return unique


# Quick test
test_recs = [
    {"id":1,"name":"Park A","lat":51.5,"lon":-0.1,"status":"active","city":"London"},
    {"id":2,"name":"park a","lat":51.5,"lon":-0.1,"status":"active","city":"London"},  # dup
    {"id":3,"name":"Park B","lat":51.6,"lon":-0.2,"status":"active","city":"London"},
]
result = deduplicate(test_recs)
print(f"
Dedup test: {len(test_recs)} → {len(result)} records  ✅")
````

````python
# 🔬 Step 5 — The pipeline orchestrator

import math as _math   # already imported above; alias to avoid re-import noise

def run_pipeline(raw_data: list[tuple]) -> tuple[list[dict], dict]:
    """Run the full Wi-Fi hotspot data-cleaning pipeline.

    Stages:
        1. Clean and validate every record.
        2. Deduplicate.
        3. Build a quality report.

    Args:
        raw_data: List of raw tuples from the data source.

    Returns:
        (clean_records, quality_report) tuple.
    """
    banner = "=" * 58

    print(banner)
    print("  Wi-Fi Hotspot Data Cleaning Pipeline — Portara City")
    print(banner)
    print(f"  Input: {len(raw_data)} records
")

    # ── Stage 1: clean + validate ─────────────────────────────
    print("--- Stage 1: Cleaning and validation ---")
    cleaned = []
    for raw in raw_data:
        rec = clean_hotspot_record(raw)
        if rec:
            cleaned.append(rec)
    n_invalid = len(raw_data) - len(cleaned)
    print(f"
  Passed: {len(cleaned)}  |  Rejected: {n_invalid}
")

    # ── Stage 2: deduplicate ──────────────────────────────────
    print("--- Stage 2: Deduplication ---")
    unique    = deduplicate(cleaned)
    n_dups    = len(cleaned) - len(unique)
    print(f"
  Kept: {len(unique)}  |  Removed: {n_dups}
")

    # ── Stage 3: quality report ───────────────────────────────
    status_counts = {}
    city_counts   = {}
    for r in unique:
        status_counts[r["status"]] = status_counts.get(r["status"], 0) + 1
        city_counts[r["city"]]     = city_counts.get(r["city"],     0) + 1

    report = {
        "total_input"       : len(raw_data),
        "invalid_rejected"  : n_invalid,
        "duplicates_removed": n_dups,
        "clean_output"      : len(unique),
        "status_breakdown"  : status_counts,
        "city_breakdown"    : city_counts,
        "acceptance_rate_%" : round(len(unique) / len(raw_data) * 100, 1),
    }

    print(banner)
    print("  Quality Report")
    print(banner)
    for k, v in report.items():
        print(f"  {k:<25}: {v}")
    print(banner)

    return unique, report


# ── Run it! ───────────────────────────────────────────────────
if __name__ == "__main__" or True:     # True forces execution in notebook
    clean_records, quality_report = run_pipeline(RAW_HOTSPOTS)
````

````python
# 🔬 Step 7 — Compute distances from city centre and visualise ranking

import math

def haversine_km(lat1, lon1, lat2, lon2) -> float:
    """Great-circle distance (km) — re-stated here for self-contained lab."""
    R = 6_371.0
    phi1, phi2 = math.radians(lat1), math.radians(lat2)
    dphi       = math.radians(lat2 - lat1)
    dlambda    = math.radians(lon2 - lon1)
    a = math.sin(dphi/2)**2 + math.cos(phi1)*math.cos(phi2)*math.sin(dlambda/2)**2
    return R * 2 * math.atan2(math.sqrt(a), math.sqrt(1 - a))


LONDON_CENTRE = (51.5074, -0.1278)     # Charing Cross, traditional centre

for r in clean_records:
    r["dist_centre_km"] = haversine_km(
        LONDON_CENTRE[0], LONDON_CENTRE[1], r["lat"], r["lon"]
    )

ranked = sorted(clean_records, key=lambda r: r["dist_centre_km"])

print("Hotspots ranked by distance from city centre:
")
print(f"  {'Rank':>4}  {'Name':<25} {'Dist (km)':>10}  Bar")
print("  " + "─" * 62)
for i, r in enumerate(ranked, 1):
    bar = "█" * int(r["dist_centre_km"] / 0.5)
    print(f"  {i:>4}  {r['name']:<25} {r['dist_centre_km']:>10.3f}  {bar}")
````

````python
# 🔬 Step 8 — Export to CSV string (portable, no file I/O needed in Colab)

def to_csv_string(records: list[dict]) -> str:
    """Convert clean records to a CSV-formatted string.

    Args:
        records: List of clean record dicts.

    Returns:
        Multi-line CSV string with header row.
    """
    FIELDS = ["id", "name", "lat", "lon", "status", "city", "dist_centre_km"]
    lines  = [",".join(FIELDS)]   # header

    for r in records:
        row = [
            str(r.get("id", "")),
            f'"{r.get("name", "")}"',        # quote name to handle commas
            f'{r.get("lat", ""):.6f}',
            f'{r.get("lon", ""):.6f}',
            r.get("status", ""),
            r.get("city",   ""),
            f'{r.get("dist_centre_km", 0):.4f}',
        ]
        lines.append(",".join(row))

    return "
".join(lines)


csv_output = to_csv_string(clean_records)
print("CSV output (first 600 chars):
")
print(csv_output[:600])
print("...
")

# Optionally write to disk (works on both Colab and local)
try:
    import pathlib
    out_path = pathlib.Path("portara_wifi_hotspots_clean.csv")
    out_path.write_text(csv_output, encoding="utf-8")
    print(f"✅ Saved to: {out_path.resolve()}")
except Exception as e:
    print(f"⚠️  Could not write file: {e} — CSV string is still available above")
````
